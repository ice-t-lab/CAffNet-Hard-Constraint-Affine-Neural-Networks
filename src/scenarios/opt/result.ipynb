# OPT Results

Build the OPT cross-seed LaTeX table from saved evaluation files. This notebook does not evaluate networks again.

```python
import sys
from pathlib import Path

import torch

REPO_ROOT = Path.cwd().resolve()
while not (REPO_ROOT / "src").is_dir() and REPO_ROOT != REPO_ROOT.parent:
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from src.utils.result import build_average_table, discover_seeds

torch.set_default_dtype(torch.float64)

RESULT_ROOT = REPO_ROOT / "results" / "opt"
print("Available runs:", sorted(path.name for path in RESULT_ROOT.glob("*") if path.is_dir()))

DIR_NAME = "final"
RUN_DIR = RESULT_ROOT / DIR_NAME
OUTPUT_DIR = RUN_DIR

# METHODS = ["Optimizer (IPOPT)", "NN", "HardNet", "CAffNet-FF", "CAffNet-TF"]
METHODS = ["Optimizer (IPOPT)", "NN", "HardNet", "CAffNet-FF", "CAffNet-FF (Lite)"]
# SEEDS = discover_seeds(RUN_DIR) if RUN_DIR.exists() else []
SEEDS = [1, 2, 5, 7, 12]

RUN_DIR
```

## Average Table Across Seeds

```python
table = build_average_table(
    RUN_DIR,
    SEEDS,
    METHODS,
    OUTPUT_DIR,
    metric_rename={
        "Max ineq viol.": ("Ineq viol.", "Max"),
        "Mean ineq viol.": ("Ineq viol.", "Mean"),
        "Num ineq viol. (%)": ("Ineq viol.", "Num (%)"),
        "Max eq viol.": ("Eq viol.", "Max"),
        "Mean eq viol.": ("Eq viol.", "Mean"),
        "Num eq viol. (%)": ("Eq viol.", "Num (%)"),
    },
    column_rename={
        "Ineq viol.": "Inequality violation",
        "Eq viol.": "Equality violation",
    },
    bold_best=True,
)
table
```
